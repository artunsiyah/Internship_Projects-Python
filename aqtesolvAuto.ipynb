```python
import pyautogui
import openpyxl
import pyperclip
import time
import os
import glob


def guileftc(xk, yk):
    pyautogui.moveTo(x=xk, y=yk)
    pyautogui.mouseDown(button='left')
    time.sleep(0.1)
    pyautogui.mouseUp(button='left')


def kutuya_tikla(dosya_ismi, px_kutu):
    resim_yolu = r"C:\Users\artun\Anaconda\Analiz\Hedefler"
    hedef_resim = os.path.join(resim_yolu, dosya_ismi)
    try:
        # Resmin ekrandaki (sol_x, ust_y, genislik, yukseklik) bilgilerini al
        kutu = pyautogui.locateOnScreen(hedef_resim, confidence=0.8)
        if kutu is not None:
            # Tik kutusu resmin en solunda yer alıyor
            # X ekseninde: Resmin en solundan 10 piksel sağa
            # Y ekseninde: Resmin dikey olarak tam ortasına
            hedef_x = kutu.left + px_kutu 
            hedef_y = kutu.top + (kutu.height / 2)
            print(f"Bulundu! Tik kutusuna tıklanıyor: X={hedef_x}, Y={hedef_y}")
            pyautogui.click(x=hedef_x, y=hedef_y)
        else:
            print("Görüntü bulunamadı.")
    except Exception as e:
        print(f"Bir hata oluştu: {e}")


# 0.1 Tüm kuyuları tutacağı ana liste
tum_kuyular = []
# Ana klasör konumu
ana_proje_dizini = (r"C:\Users\artun\Desktop\STAJYER"
                    r"\SondajFaaliyetKutukPmpjTst")
girdi = os.path.join(ana_proje_dizini, "DosyaNo", "KuyuNo")
dosya = os.path.join(girdi, "*_PT.xlsx")
bulunan_dosyalar = glob.glob(dosya)

# 1. AŞAMA: VERİ OKUMA VE TOPLAMA
if bulunan_dosyalar:
    hedef_dosya = bulunan_dosyalar[0]

    wb = openpyxl.load_workbook(hedef_dosya, data_only=True)
    sheet = wb["Düşüm"]
    dosya_adi_ham, _ = os.path.splitext(os.path.basename(hedef_dosya))
    zamanla_dusum = ""
    baslangic_satiri = 3

    for row in range(baslangic_satiri, sheet.max_row + 1):
        zaman_degeri = sheet[f'A{row}'].value
        dusum_degeri = sheet[f'C{row}'].value
        # FİLTRE: C sütunu boşsa (None) veya 0 ise döngüyü sonlandır
        if dusum_degeri in [None, "", 0, "0"]:
            print(f"C{row} hücresinde veri bitti. Okuma {row-1}. satırda durduruldu.")
            break
        # ONDALIK YUVARLAMA (Sadece sayısal değerler için çalışır)
        # Düşüm değerini virgülden sonra 3, Zaman değerini 1 basamağa yuvarlıyoruz
        if isinstance(dusum_degeri, (float, int)):
            dusum_degeri = round(dusum_degeri, 3)
        if isinstance(zaman_degeri, (float, int)):
            zaman_degeri = round(zaman_degeri, 1)
        # A sütunu güvenlik
        if zaman_degeri is None:
            zaman_degeri = ""
        # A ve C sütununu aralarında TAB ("\t") olacak şekilde birleştirir
        zamanla_dusum += f"{zaman_degeri}\t{dusum_degeri}\n"

    # D ve L değerlerini hücreden çeker
    d_kapali_ham = sheet["F8"].value
    l_filtre_ham = sheet["F9"].value

    # Eğer değerler sayısal ise en yakın tam sayıya yuvarla, değilse olduğu 
    # gibi bırakır (None vb. durumlar için)
    d_kapali_tam = (
        round(d_kapali_ham, 2)
        if isinstance(d_kapali_ham, (float, int))
        else d_kapali_ham
    )
    l_filtre_tam = (
        round(l_filtre_ham, 2)
        if isinstance(l_filtre_ham, (float, int))
        else l_filtre_ham
    )
    # Parametreleri bir sözlük (dictionary) formatında paketler
    kuyu_verisi = {
        "Kuyu_Adi": dosya_adi_ham.replace("_PT", ""),
        "Debi": sheet['E1'].value,
        "Kalinlik": sheet['F7'].value,
        "D_Kapali": d_kapali_tam,
        "L_Filtre": l_filtre_tam,
        "R_Casing": sheet["U2"].value,
        "W_Borehole": sheet["S2"].value
    }
    tum_kuyular.append(kuyu_verisi)  # Paketi listeye ekle
# Excel verisi alındıktan dosyayı kapat
# Alınan parametreleri çıktı olarak çıkar
    wb.close()
    print(f"Kuyu Adı:{kuyu_verisi['Kuyu_Adi']}")
    print(f"Debi:{kuyu_verisi['Debi']}")
    print(f"B_kalınlık:{kuyu_verisi['Kalinlik']}")
    print(f"D:{kuyu_verisi['D_Kapali']}")
    print(f"Filtre:{kuyu_verisi['L_Filtre']}")
    print(f"Borehole_Yarıçap:{kuyu_verisi['W_Borehole']}")
    print(f"R_Yarıçap:{kuyu_verisi['R_Casing']}")


# Fareyi götüreceği kordinatlar
pyautogui.moveTo(x=1662, y=27, duration=0.5)

# Tıklama eylemini milisaniye seviyesinde "basılı tutarak" yap
pyautogui.mouseDown()
time.sleep(0.1)  # Algıma Süresi
pyautogui.mouseUp()
time.sleep(1)
pyautogui.moveTo(x=1734, y=188)
time.sleep(1)
pyautogui.moveTo(x=1564, y=188)
pyautogui.mouseDown()
time.sleep(0.1)
pyautogui.mouseUp()
#  Yeni Dosya Oluşturma(dosya süresi 10s)
time.sleep(7)
pyautogui.moveTo(x=18, y=124)
pyautogui.mouseDown()
time.sleep(0.1)
pyautogui.mouseUp()
# Menü tercihleri
pyautogui.press("tab", presses=2)
pyautogui.press("down")
pyautogui.press("enter")
time.sleep(0.5)
# Test Tipi
pyautogui.press("down", presses=2)
pyautogui.press("enter")
time.sleep(0.5)
#  Project İnfo, Aquifer Data,Well Data, Well Constraction, Well Radius,
#  Pumping Rates,
for kuyu in tum_kuyular:
    pyautogui.press("enter", presses=2)
    pyautogui.press("tab", presses=4)
    pyautogui.write(str(kuyu["Kuyu_Adi"]))
    pyautogui.press("tab", presses=6)
    pyautogui.press("enter")
    time.sleep(0.5)
    pyautogui.write(str(kuyu["Kalinlik"]))
    pyautogui.press("tab", presses=4)
    pyautogui.press("enter")
    time.sleep(0.5)
    pyautogui.press("tab", presses=4)
    pyautogui.press("enter")
    time.sleep(0.5)
    if kuyu['D_Kapali'] in [0, "0", None] and kuyu['L_Filtre'] in [0, "0", None]:
        pyautogui.press("tab", presses=3)
        pyautogui.press("enter")
    else:
        pyautogui.press("down")
        pyautogui.press("tab")
        pyautogui.write(str(kuyu["D_Kapali"]))
        pyautogui.press("tab")
        pyautogui.write(str(kuyu['L_Filtre']))
        pyautogui.press("tab", presses=3)
        pyautogui.press("enter")
        time.sleep(0.5)
    pyautogui.write(str(kuyu['R_Casing']))
    pyautogui.press("tab")
    pyautogui.write(str(kuyu['R_Casing']))
    pyautogui.press("tab")
    pyautogui.write(str(kuyu['W_Borehole']))
    pyautogui.press("tab", presses=2)
    pyautogui.press("enter")
    pyautogui.write("0")
    pyautogui.press("tab")
    pyautogui.press("right")
    pyautogui.write(str(kuyu['Debi']))
    pyautogui.moveTo(x=980, y=740)
    pyautogui.mouseDown()
    time.sleep(0.1)
    pyautogui.mouseUp()
    pyperclip.copy(zamanla_dusum)
    pyautogui.moveTo(x=1186, y=497)
    pyautogui.mouseDown()
    time.sleep(0.1)
    pyautogui.mouseUp()
    pyautogui.press("enter")
    time.sleep(1)
    pyautogui.press("enter")


# Kayıt
pyautogui.moveTo(x=18, y=88)
time.sleep(0.5)
pyautogui.mouseDown()
time.sleep(0.1)
pyautogui.mouseUp()
pyautogui.moveTo(x=85, y=233)
pyautogui.mouseDown()
time.sleep(0.1)
pyautogui.mouseUp()
tam_kayit_yolu = os.path.join(girdi, f"{tum_kuyular[0]['Kuyu_Adi']}.aqt")
pyperclip.copy(tam_kayit_yolu)
time.sleep(0.5)
pyautogui.hotkey("ctrl", "v")
pyautogui.press("enter")

# Çözümlemeye Hazırlık
guileftc(393, 123)
guileftc(333, 155)
guileftc(615, 123)
guileftc(652, 123)
time.sleep(1)
kutuya_tikla("bourdet.png", 10)
pyautogui.press("tab")
pyautogui.write("0.8")
pyautogui.press("tab")
pyautogui.press("enter")
guileftc(1550, 168)
pyautogui.hotkey("ctrl", "s")
pyautogui.rightClick(x=1040, y=240)
pyautogui.press("up", presses=5)
pyautogui.press("enter")
time.sleep(1)
kutuya_tikla("inactive.png", 10)
```
