```python
import ctypes
import time

# Windows'a ekranında fare'nin kordinatlarını görmek için
try:
    ctypes.windll.shcore.SetProcessDpiAwareness(2)
except:
    ctypes.windll.user32.SetProcessDPIAware()


class POINT(ctypes.Structure):
    _fields_ = [("x", ctypes.c_long), ("y", ctypes.c_long)]


print("Windows API Koordinat okuyucu başlatıldı. Kapatmak için Ctrl+C yapın.")
try:
    while True:
        pt = POINT()
        ctypes.windll.user32.GetCursorPos(ctypes.byref(pt))
        # Rakamların karışmaması için sabit boşluk bıraktık
        print(f"Fare Koordinatları: X: {pt.x:<5} | Y: {pt.y:<5}", end="\r")
        time.sleep(0.1)
except KeyboardInterrupt:
    print("\nÇıkış yapıldı.")
```
